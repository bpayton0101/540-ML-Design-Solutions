**Section 1**

**a) Inference Network: Project Bidding Expert System**

The network follows a Forward Chaining logic, flowing from initial facts to intermediate conclusions, and finally to the terminal goal.

**Network Architecture & Layers**
**Initial Facts (Evidence Layer):**

* Order Backlog and Labor Supply: These two inputs feed directly into the first layer of logic (R1 and R2).

* Customer Importance: This input remains "dormant" until an intermediate conclusion is reached.

* Historical Data: This is the final "filter" used to determine the methodology.

**Intermediate Conclusions (Inference Layer):**

* Tentative Decision: This is a "hidden" attribute. It is not an initial fact, but it must be established by R2 before the system can evaluate the customer's importance.

* Rules R1, R2, & R3: These represent the first "wave" of the inference engine.

**Terminal Decisions (Goal Layer):**

* Bid Decision: This is the pivotal conclusion. If R3 or R6 fire, the decision is Yes. If R1 or R7 fire, the decision is Decline.

* Bid Methodology: The final output. If the bid is "Yes," the system selects GP (R4) or HRS (R5). If "Decline," it selects NA (R8).

**Section 2**

**b) Rule/Premise Property Table**

This table represents the Knowledge Base (KB). Each rule is broken down by its constituent clauses, defining the attribute being tested and the value required to trigger the rule’s conclusion.

**c) Solve the rule base of Problem #1 using the facts given in Problem #2 by applying Forward Chaining (start inference with order backlog = large)**

```python
import pandas as pd
from IPython.display import display, HTML

# ==========================================
# PART B: RULE/PREMISE PROPERTY TABLE
# ==========================================
properties_data = [
    {"Rule ID": "R1", "Clause": 1, "Clause Attribute": "Order Backlog", "Clause Value": "Large", "Conclusion Attribute": "Bid Decision", "Conclusion Value": "Decline"},
    {"Rule ID": "R1", "Clause": 2, "Clause Attribute": "Labor Supply", "Clause Value": "Nil", "Conclusion Attribute": "", "Conclusion Value": ""},
    {"Rule ID": "R2", "Clause": 1, "Clause Attribute": "Order Backlog", "Clause Value": "Large", "Conclusion Attribute": "Tentative Decision", "Conclusion Value": "Consider"},
    {"Rule ID": "R2", "Clause": 2, "Clause Attribute": "Labor Supply", "Clause Value": "Adequate", "Conclusion Attribute": "", "Conclusion Value": ""},
    {"Rule ID": "R3", "Clause": 1, "Clause Attribute": "Order Backlog", "Clause Value": "Minimal/Modest", "Conclusion Attribute": "Bid Decision", "Conclusion Value": "Yes"},
    {"Rule ID": "R4", "Clause": 1, "Clause Attribute": "Bid Decision", "Clause Value": "Yes", "Conclusion Attribute": "Bid Methodology", "Conclusion Value": "GP"},
    {"Rule ID": "R4", "Clause": 2, "Clause Attribute": "Historical Data", "Clause Value": "Available", "Conclusion Attribute": "", "Conclusion Value": ""},
    {"Rule ID": "R5", "Clause": 1, "Clause Attribute": "Bid Decision", "Clause Value": "Yes", "Conclusion Attribute": "Bid Methodology", "Conclusion Value": "HRS"},
    {"Rule ID": "R5", "Clause": 2, "Clause Attribute": "Historical Data", "Clause Value": "Not Available", "Conclusion Attribute": "", "Conclusion Value": ""},
    {"Rule ID": "R6", "Clause": 1, "Clause Attribute": "Tentative Decision", "Clause Value": "Consider", "Conclusion Attribute": "Bid Decision", "Conclusion Value": "Yes"},
    {"Rule ID": "R6", "Clause": 2, "Clause Attribute": "Customer", "Clause Value": "Important", "Conclusion Attribute": "", "Conclusion Value": ""},
    {"Rule ID": "R7", "Clause": 1, "Clause Attribute": "Tentative Decision", "Clause Value": "Consider", "Conclusion Attribute": "Bid Decision", "Conclusion Value": "Decline"},
    {"Rule ID": "R7", "Clause": 2, "Clause Attribute": "Customer", "Clause Value": "Not Important", "Conclusion Attribute": "", "Conclusion Value": ""},
    {"Rule ID": "R8", "Clause": 1, "Clause Attribute": "Bid Decision", "Clause Value": "Decline", "Conclusion Attribute": "Bid Methodology", "Conclusion Value": "NA"}
]

print("<h2>b) Rule/Premise Property Table</h2>")
df_properties = pd.DataFrame(properties_data)
display(HTML(df_properties.to_html(index=False)))

# ==========================================
# PART C: FORWARD CHAINING INFERENCE TRACE
# ==========================================
print("<hr><h2>c) Forward Chaining Inference Trace</h2>")

# Initial Facts from Problem #2
working_memory = {
    "Order Backlog": "Large",
    "Labor Supply": "Adequate",
    "Customer": "Important",
    "Historical Data": "Available"
}

# Knowledge Base for the Inference Engine
rules = [
    {"id": "R1", "premise": {"Order Backlog": "Large", "Labor Supply": "Nil"}, "attr": "Bid Decision", "val": "Decline"},
    {"id": "R2", "premise": {"Order Backlog": "Large", "Labor Supply": "Adequate"}, "attr": "Tentative Decision", "val": "Consider"},
    {"id": "R3", "premise": {"Order Backlog": "Minimal"}, "attr": "Bid Decision", "val": "Yes"},
    {"id": "R4", "premise": {"Bid Decision": "Yes", "Historical Data": "Available"}, "attr": "Bid Methodology", "val": "GP"},
    {"id": "R5", "premise": {"Bid Decision": "Yes", "Historical Data": "Not Available"}, "attr": "Bid Methodology", "val": "HRS"},
    {"id": "R6", "premise": {"Tentative Decision": "Consider", "Customer": "Important"}, "attr": "Bid Decision", "val": "Yes"},
    {"id": "R7", "premise": {"Tentative Decision": "Consider", "Customer": "Not Important"}, "attr": "Bid Decision", "val": "Decline"},
    {"id": "R8", "premise": {"Bid Decision": "Decline"}, "attr": "Bid Methodology", "val": "NA"}
]

attr_queue = list(working_memory.keys())
fired_rules = []
iteration = 0

# Initial State
print(f"<h3>Iteration {iteration}: Initial State</h3>")
display(pd.DataFrame(list(working_memory.items()), columns=["Attribute", "Value"]))
print(f"<b>Attribute Queue:</b> {attr_queue}")

# Inference Loop
inferred = True
while inferred:
    inferred = False
    for rule in rules:
        if rule["id"] in fired_rules: continue

        # Check premises
        if all(working_memory.get(k) == v for k, v in rule["premise"].items()):
            iteration += 1
            inferred = True
            fired_rules.append(rule["id"])
            working_memory[rule["attr"]] = rule["val"]
            attr_queue.insert(0, rule["attr"]) # Add new fact to queue

            print(f"<br><hr><h3>Iteration {iteration}: Rule {rule['id']} Fired</h3>")
            print(f"<b>Action:</b> Inferred {rule['attr']} = {rule['val']}")

            # Display Working Memory
            print("<h4>Updated Working Memory Table</h4>")
            display(pd.DataFrame(list(working_memory.items()), columns=["Attribute", "Value"]))

            # Display Attribute Queue
            print(f"<b>Current Attribute Queue:</b> {attr_queue}")
            break # Restart scan with new fact

print(f"<br><div style='background-color:#e1f5fe; padding:10px;'><b>Final Goal Reached:</b> Bid Methodology = {working_memory.get('Bid Methodology')}</div>")
```


**Section 3:  Conclusion and Interpretation of Results**

The forward-chaining execution successfully navigated the project bidding rule set to reach the terminal goal of Bid Methodology = GP (Goal Programming). The inference followed a deterministic path of R2 → R6 → R4, validating the following structural logic:

**Logical Dependency:** The system correctly identified that R6 could not be evaluated until R2 provided the intermediate "Tentative Decision." This demonstrates the effectiveness of the Attribute Queue in managing data dependencies.

**Conflict Resolution:** By bold text starting the inference with Order Backlog = Large, the system ignored the "Yes" path of R3 (Minimal/Modest) and instead engaged the "Consider" logic. Despite the initial "Large" backlog, the "Adequate" labor supply prevented an immediate "Decline" (R1), allowing the system to proceed to a nuanced evaluation of customer importance.

**Final Outcome:** The convergence on GP methodology was driven by the intersection of a "Yes" bid decision and the "Available" state of historical data. If the historical data had been "Not Available," the system would have dynamically pivoted to HRS (R5), demonstrating the flexibility of a well-defined Knowledge Base.
